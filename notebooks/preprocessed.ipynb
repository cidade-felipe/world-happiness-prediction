# Traduzindo colunas

```python
from googletrans import Translator

translator = Translator()
colunas_traduzidas = {}

for col in df.columns:
   try:
      traducao = translator.translate(col, src='en', dest='pt').text
      colunas_traduzidas[col] = traducao
   except Exception:
      colunas_traduzidas[col] = col

df = df.rename(columns=colunas_traduzidas)

print("Colunas traduzidas:")
print(df.columns.tolist())

# Salvar o CSV traduzido
df.to_csv('../data/processed/world_happiness_report_colunas_pt.csv', index=False)
```

## Analisando as colunas

### Removendo as colunas dispensáveis

```python
df_2 = df_2.drop(columns=['Indicador Regional'])
df_2
```

### Renomeando as colunas

```python
df_2 = df_2.rename(columns={
   'Nome do país': 'País',
   'Escada da Vida': 'Índice de Felicidade',
   'Registrar PIB per capita' : 'PIB per capita (US$)', 
   'Expectativa de vida saudável ao nascer': 'Expectativa de vida',
   'Liberdade para fazer escolhas de vida': 'Liberdade de escolhas',   
})
df_2
```

```python
df_2.isna().sum()
```

### Preenchendo os dados NaN de cada país

```python
for coluna in df_2.columns:
   if coluna != "País":
      df_2[coluna] = df_2.groupby("País")[coluna].transform(
         lambda x: x.fillna(x.mean())
      )

```

```python
df_2.isna().sum()
```

```python
df_2 = df_2.fillna(df_2.mean(numeric_only=True))
```
