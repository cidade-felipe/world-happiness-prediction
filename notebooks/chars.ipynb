# Importando Libs e lendo o arquivo

```python
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display


df_test = pd.read_csv('../chars/data/dataset_comparacao_modelos.csv')
df_test
```

# Gráficos

## Comparativo de previsões com índice real

```python

# Lista de países para o dropdown
paises = sorted(df_test["País"].unique())
dropdown_pais = widgets.Dropdown(options=paises, value=paises[0], description="País")

def plot_modelos(pais):
   dados = df_test[df_test["País"] == pais]
   
   # média por garantia, caso tenha mais de um ano por país
   valores = dados[["Índice Real", "LR Previsto", "RF Previsto", "XGB Previsto"]].mean()
   modelos = valores.index.tolist()
   numeros = valores.values
   
   plt.figure(figsize=(6,4))
   plt.bar(modelos, numeros)
   plt.ylim(0, 10)
   plt.ylabel("Índice de Felicidade")
   plt.title(f"Comparação de Previsões para {pais}")
   plt.xticks(rotation=15)
   plt.tight_layout()
   plt.show()

widgets.interact(plot_modelos, pais=dropdown_pais);
```

## Evolução da felicidade ao longo dos anos

### Lendo o Dataset

```python
df = pd.read_csv('../data/processed/world_happiness_report_colunas_pt_preprocessed.csv')

df_evolucao = df.copy()
df_evolucao
```

```python


# Dropdown de países
paises = sorted(df_evolucao["País"].unique())
dropdown_pais = widgets.Dropdown(options=paises, value=paises[0], description="País")

def plot_evolucao(pais):
   # filtra o país
   dados = df_evolucao[df_evolucao["País"] == pais].copy()
   
   # garante que está ordenado por ano e, se tiver mais de uma linha por ano, tira a média
   dados = (
      dados.groupby("Ano", as_index=False)["Índice de Felicidade"]
      .mean()
      .sort_values("Ano")
   )

   plt.figure(figsize=(7,4))
   plt.plot(dados["Ano"], dados["Índice de Felicidade"], marker="o")
   plt.title(f"Evolução do Índice de Felicidade em {pais}")
   plt.xlabel("Ano")
   plt.ylabel("Índice de Felicidade")
   plt.ylim(0, 10)
   plt.grid(alpha=0.3)
   plt.xticks(rotation=45)
   plt.tight_layout()
   plt.show()

widgets.interact(plot_evolucao, pais=dropdown_pais);
```

```python
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Defina aqui a coluna de nome de país no shapefile
country_col = "ADMIN"  # troque se for NAME ou outra

# Dataset de felicidade: País, Ano, Índice de Felicidade
df_mapa = df_mapa.copy()
df_mapa = df_mapa.groupby(["País", "Ano"], as_index=False)["Índice de Felicidade"].mean()

# Dropdown de anos
anos = sorted(df_mapa["Ano"].unique())
dropdown_ano = widgets.Dropdown(options=anos, value=anos[-1], description="Ano")

def plot_mapa(ano):
   dados_ano = df_mapa[df_mapa["Ano"] == ano]

   # Join entre shapefile e dados (atenção no country_col)
   mapa = world.merge(dados_ano, left_on=country_col, right_on="País", how="left")

   plt.figure(figsize=(14, 6))
   ax = plt.gca()
   ax.set_title(f"Índice Médio de Felicidade Mundial – {ano}", fontsize=14)
   mapa.plot(
      column="Índice de Felicidade",
      cmap="viridis",
      linewidth=0.3,
      edgecolor="black",
      legend=True,
      missing_kwds={"color": "lightgray", "label": "Sem dados"},
      ax=ax
   )
   ax.set_axis_off()
   plt.show()

widgets.interact(plot_mapa, ano=dropdown_ano);
```
